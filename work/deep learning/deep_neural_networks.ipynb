```python
from tensorflow.keras import Sequential, layers

INPUT_SHAPE, HIDDEN_SHAPE_1, HIDDEN_SHAPE_2, OUTPUT_SHAPE = [2], 4, 3, 1
ReLU = 'relu'

model = Sequential([
    layers.Input(shape=INPUT_SHAPE),
    # the hidden ReLU layers
    layers.Dense(units=HIDDEN_SHAPE_1, activation=ReLU),
    layers.Dense(units=HIDDEN_SHAPE_2, activation=ReLU),
    # the linear output layer 
    layers.Dense(units=OUTPUT_SHAPE),
])
```
