```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
plt.rc('figure', autolayout=True)
plt.rc('axes', labelweight='bold', labelsize='large',
       titleweight='bold', titlesize=18, titlepad=10)
```

```python
import pandas as pd

red_wine = pd.read_csv('../datasets/dl-course-data/red-wine.csv')
red_wine.head()
```

```python
red_wine.shape
```

```python
input_shape = [red_wine.shape[1] - 1]
input_shape
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential, layers

model = Sequential([
    layers.Input(shape=input_shape),
    layers.Dense(1)
])
```

```python
# Get Model weights
w, b = model.weights
print(f"w={w}b, b={b}")
```

```python
model = Sequential([
    layers.Input(shape=(1,)),
    layers.Dense(1),
])

X = tf.linspace(-1.0, 1.0, 100)
y = model.predict(X)

plt.figure(dpi=100)
plt.plot(X, y, 'k')
plt.xlim(-1, 1)
plt.ylim(-1, 1)
plt.xlabel("Input: x")
plt.ylabel("Target y")

w, b = model.weights
plt.title(f"Weight: {w[0][0]:0.2f}\nBias: {b[0]:0.2f}")
plt.show()
```
