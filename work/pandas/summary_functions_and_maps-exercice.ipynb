```python
import pandas as pd

pd.set_option("display.max_rows", 5)

# Read data
reviews = pd.read_csv("../datasets/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
reviews.head()
```

```python
median_points = reviews.points.median()
median_points
```

```python
countries = reviews.country.unique()
countries
```

```python
reviews_per_country = reviews.country.value_counts()
reviews_per_country
```

```python
# price column with the mean price subtracted
mean_price = reviews.price.mean()
centered_price = reviews.price - mean_price
centered_price
```

```python
# title of the wine with the highest points-to-price ratio
bargain_wine = reviews.loc[(reviews.points / reviews.price).idxmax(), 'title']
bargain_wine
```

```python
search_words = ['fruity', 'tropical']
count_words = [reviews.description.map(lambda d: w in d).sum() for w in search_words]
descriptor_counts = pd.Series(count_words, index=search_words)
descriptor_counts
```

```python
# equivalent
search_words = ['fruity', 'tropical']
descriptor_counts = pd.Series([reviews.description.str.contains(w).sum() for w in search_words], 
                              index=['tropical', 'fruity'])
descriptor_counts
```

```python
star_ratings = reviews.loc[:,['country', 'points']].apply(
    lambda row: 3 if (row['country'] == 'Canada' or row['points'] >= 95) 
    else 2 if row['points'] >= 85 else 1, axis=1
)
star_ratings
```
