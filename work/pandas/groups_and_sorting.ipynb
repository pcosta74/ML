```python
import pandas as pd
pd.set_option("display.max_rows", 5)

reviews = pd.read_csv("../datasets/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
reviews.head()
```

```python
# value_counts() equivalent
reviews.groupby('points').points.count()
```

```python
# get the cheapest wine in each point value category
reviews.groupby('points').price.min()
```

```python
# select the name of the first wine reviewed from each winery 
reviews.groupby('winery').title.first()
```

```python
# Select the best wine by country and province:
group_cols = ['country', 'province']
reviews.loc[reviews.groupby(group_cols).points.idxmax()].set_index(group_cols)
```

```python
# generate a simple statistical summary
reviews.groupby(['country']).price.agg([len, "min", "max"])
```

```python
# Multi-indices
countries_reviewed = reviews.groupby(['country', 'province']).description.agg([len])
countries_reviewed
```

```python
mi = countries_reviewed.index
type(mi)
```

```python
countries_reviewed.reset_index()
```

```python
# Sorting by "len"
countries_reviewed = countries_reviewed.reset_index()
countries_reviewed.sort_values(by='len')
```

```python
# Sorting by "len" descending
countries_reviewed.sort_values(by='len', ascending=False)
```

```python
# sort by index values
countries_reviewed.sort_index()
```

```python
# sort by more than 12 column
countries_reviewed.sort_values(by=['country', 'len'])
```
