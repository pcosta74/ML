```python
# Pass the entries to select as a list
reviews.iloc[[0, 1, 2], 0]
```

```python
# Select the last five rows of the dataset
reviews.iloc[-5:]
```

```python
# Get the very first entry in reviews (first colummn, first row (which is 'country'))
reviews.loc[0, 'country']
```

```python
# Select all rows for 'taster_name', 'taster_twitter_handle', and 'points' columns
reviews.loc[:, ['taster_name', 'taster_twitter_handle', 'points']]
```

```python
# Set index to the 'title' field
reviews.set_index("title")
```

```python
# Checking if each wine is Italian or not
reviews.country == 'Italy'
```

```python
# Select italian wines
reviews.loc[reviews.country == 'Italy']
```

```python
# Select italian wines that accrued at least 90 points
reviews.loc[(reviews.country == 'Italy') & (reviews.points >= 90)]
```

```python
# Select wines that are eitehr italian OR accrued at least 90 points
reviews.loc[(reviews.country == 'Italy') | (reviews.points >= 90)]
```

```python
# Select italian or french wines
reviews.loc[reviews.country.isin(['Italy', 'France'])]
```

```python
# Select wines whose price is not empty (NaN)
reviews.loc[reviews.price.notnull()]
```

```python
# Assigning constant data to a column
reviews['critic'] = 'everyone'
reviews['critic']
```

```python
# Assigning an iterable to a column
reviews['index_backwards'] = range(len(reviews), 0, -1)
reviews['index_backwards']
```
