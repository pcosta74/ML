```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

# Load the data
file_path = '../datasets/home-data-for-ml-course/train.csv'
home_data = pd.read_csv(file_path)

# Select columns corresponding to features and target
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']
X = home_data[features]
y = home_data.SalePrice

X.head()
```

```python
# Split into validation and training data
X_train, X_valid, y_train, y_valid = train_test_split(X, y, random_state=1)

# Define a random forest model
model = RandomForestRegressor(random_state=1)
model.fit(X_train, y_train)
pred_valid = model.predict(X_valid)
valid_mae = mean_absolute_error(y_valid, pred_valid)

print("Validation MAE for Random Forest Model: {:,.0f}".format(valid_mae))
```

```python
# Create and fit a new model
model = RandomForestRegressor(max_leaf_nodes=100, random_state=1)
model.fit(X, y)
```

```python
# Load test data
file_path = '../datasets/home-data-for-ml-course/test.csv'
test_data = pd.read_csv(file_path)

# create X_test which comes from test_data but includes only the columns you used for prediction.
X_test = test_data[features]

# make predictions which we will submit. 
y_test = model.predict(X_test)
```

```python
# Run the code to save predictions in the ".csv" format
output = pd.DataFrame({'Id': test_data.Id, 'SalePrice': y_test})
# output.to_csv('submission.csv', index=False)
```
