```python
import pandas as pd

# Load data
file_path = "../datasets/melbourne-housing-snapshot/melb_data.csv"
melbourne_data = pd.read_csv(file_path)

# Filter rows with missing values
melbourne_data.dropna(axis=0, inplace=True)
```

```python
# Choose target and features
features = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea', 'YearBuilt', 'Lattitude', 'Longtitude']
X = melbourne_data[features]
y = melbourne_data.Price
```

```python
# split data into training and validation data, for both features and target
from sklearn.model_selection import train_test_split
train_X, val_X, train_y, val_y = train_test_split(X, y,random_state = 0)
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error

def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return (mae)
```

```python
# compare MAE with differing values of max_leaf_nodes
for leaf_nodes in [5, 50, 500, 5000]:
    mae = get_mae(leaf_nodes, train_X, val_X, train_y, val_y)
    print(f"Max leaf nodes: {leaf_nodes}  \t\t Mean Absolute Error:  {mae}")
```
