```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

pd.plotting.register_matplotlib_converters()
%matplotlib inline
```

```python
# Load data
cancer_data = pd.read_csv("../datasets/data-for-datavis/cancer.csv", index_col='Id')
cancer_data.head()
```

```python
# In the first five rows of the data, what is the largest value for 'Perimeter (mean)'?
max_perim = cancer_data.head()['Perimeter (mean)'].max()
max_perim
```

```python
# What is the value for 'Radius (mean)' for the tumor with Id 8510824?
mean_radius = cancer_data.head().loc[8510824, 'Radius (mean)']
mean_radius
```

```python
# Histograms for benign and maligant tumors
sns.histplot(data=cancer_data, x='Area (mean)', hue='Diagnosis')
```

```python
from IPython.display import display, Markdown

content = f"""
Based on the histograms above

**Q**: **Do malignant tumors have higher or lower values for `Area (mean)` (relative to benign tumors), on average?**<br>
**A**: Malignant tumors have higher values for `Area (mean)`, on average`.

**Q**: **Which tumor type seems to have a larger range of potential values?**<br>
**A**: Malignant tumors have a larger range of potential values.
"""

display(Markdown(content))

```

```python
sns.kdeplot(data=cancer_data, x='Radius (worst)', hue='Diagnosis', fill=True)
```
