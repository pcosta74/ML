```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

pd.plotting.register_matplotlib_converters()
%matplotlib inline

# Load data
ign_data = pd.read_csv("../datasets/data-for-datavis/ign_scores.csv", index_col="Platform")
```

```python
ign_data.head()
```

```python
ign_data.tail()
```

```python
# What is the highest average score received by PC games, for any genre?
high_score = ign_data.loc['PC'].max()
high_score
```

```python
# On the Playstation Vita platform, which genre has the lowest average score?
worst_genre = ign_data.loc['PlayStation Vita'].idxmin()
worst_genre
```

```python
plt.figure(figsize=(10,6))
plt.title("Average score for racing games by platform")
sns.barplot(x=ign_data.Racing, y=ign_data.index)
```

```python
plt.figure(figsize=(10,6))
plt.title("Average score for Game type, by Platform")
sns.heatmap(data=ign_data, annot=True)
```

```python
# Find the maximum value in the entire DataFrame
max_value = ign_data.max().max()

# Find the index (row name) where the maximum value occurs
row_name = ign_data.max(axis=1).idxmax()

# Find the column where the maximum value occurs
col_name = ign_data.max(axis=0).idxmax()

print(f"Max value: {max_value}")
print(f"- {row_name} x {col_name}")
```

```python
# Find the maximum value in the entire DataFrame
min_value = ign_data.min().min()

# Find the positions in which the minimum value occurs
min_positions = ign_data[ign_data == min_value].stack().index

print(f"Max value: {max_value}")
for row_name, col_name in min_positions:
    print(f"- {row_name} x {col_name}")
```
