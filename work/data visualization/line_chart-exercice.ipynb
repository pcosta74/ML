```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

pd.plotting.register_matplotlib_converters()
%matplotlib inline

# Load data
museum_data = pd.read_csv("../datasets/data-for-datavis/museum_visitors.csv",
                          index_col="Date", parse_dates=True)
```

```python
# How many visitors did the Chinese American Museum receive in July 2018?
ca_museum_jul18 = museum_data['Chinese American Museum']['2018-07-01']
ca_museum_jul18
```

```python
# How many more visitors did Avila Adobe receive than the Firehouse Museum?
avila_oct18 = museum_data['Avila Adobe']['2018-10-01'] - museum_data['Firehouse Museum']['2018-10-01']
avila_oct18
```

```python
plt.figure(figsize=(16,6))
sns.lineplot(data=museum_data)
plt.title("Monthly visitors for the museums in the City of Los Angeles in 2014-2018")
plt.xlabel("Date")
```

```python
plt.figure(figsize=(16,6))
sns.lineplot(data=museum_data['Avila Adobe'])
plt.title("Monthly visitors for the Avila Adobe museum in 2014-2018")
plt.xlabel("Date")
```
