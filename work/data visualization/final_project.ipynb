```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.plotting.register_matplotlib_converters()
%matplotlib inline
```

```python
data_path = "../datasets/fivethirtyeight-comic-characters-dataset/dc-wikia-data.csv"
comics_data = pd.read_csv(data_path)
comics_data.head()
```

```python
sns.histplot(data=comics_data, x='ALIGN', hue='SEX')
```

```python
sns.lmplot(x="YEAR", y="APPEARANCES", hue="ALIGN", data=comics_data)
```
