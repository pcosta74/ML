```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.plotting.register_matplotlib_converters()
%matplotlib inline

# Load data
spotify_data = pd.read_csv("../datasets/data-for-datavis/spotify.csv", index_col="Date", parse_dates=True)
spotify_data.tail()
```

```python
# Change the style of the figure
sns.set_style("dark")

# Line chart 
plt.figure(figsize=(16,6))
sns.lineplot(data=spotify_data)
```

```python
styles = ["darkgrid", "whitegrid", "dark", "white", "ticks"]

# Change the style of the figure
for s in styles:
    sns.set_style(s)
    plt.figure(figsize=(16,6))
    sns.lineplot(data=spotify_data)
```
