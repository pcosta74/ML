```python
import pandas as pd

# Read the data
data = pd.read_csv('../datasets/melbourne-housing-snapshot/melb_data.csv')

# Select subset of predictors
features = ['Rooms', 'Distance', 'Landsize', 'BuildingArea', 'YearBuilt']
X = data[features]
y = data.Price
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

pipeline = Pipeline(steps=[
    ('preprocessor', SimpleImputer()),
    ('model', RandomForestRegressor(n_estimators=50, random_state=0))
])
```

```python
from sklearn.model_selection import cross_val_score

# Multiply by -1 since sklearn calculates *negative* MAE
scores = -1 * cross_val_score(pipeline, X, y,
                              cv=5,
                              scoring='neg_mean_absolute_error')

print(f"MAE scores: {scores}")
print(f"Average MAE score (across experiments): {scores.mean()}")
```
