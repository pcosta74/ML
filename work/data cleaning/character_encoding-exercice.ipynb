```python
import numpy as np
import pandas as pd
import charset_normalizer

np.random.seed(0)
```

```python
sample_entry = b'\xa7A\xa6n'
print(sample_entry)
print('data type:', type(sample_entry))
```

```python
# create a variable new_entry that changes the encoding from "big5-tw" to "utf-8".
# new_entry should have the bytes datatype.
new_entry = sample_entry.decode('big5-tw').encode()
print(new_entry)
print(new_entry.decode())
```

```python
filename = "../datasets/fatal-police-shootings-in-the-us/PoliceKillingsUS.csv"

with open(filename, 'rb') as rawdata:
    result = charset_normalizer.detect(rawdata.read())
print(result)

police_killings_data = pd.read_csv(filename, encoding=result['encoding'])
police_killings_data.head()
```
