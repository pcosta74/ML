```python
import numpy as np
import pandas as pd

# helpful character encoding module
import charset_normalizer

np.random.seed(0)
```

```python
before = "This is the euro symbol: €"
type(before)
```

```python
after = before.encode("utf-8", errors="replace")
type(after)
```

```python
print(after.decode("utf-8"))
# print(after.decode("ascii")) <-- raises a UnicodeDecodeError
```

```python
before = "This is the euro symbol: €"
after = before.encode("ascii", errors = "replace")
print(after.decode("ascii"))
```

```python
filename = "../datasets/kickstarter-projects/ks-projects-201612.csv"
# ks2016_data = pd.read_csv(filename) <-- raises a UnicodeDecodeError
```

```python
# look at the first ten thousand bytes to guess the character encoding
with open(filename, 'rb') as rawdata:
    result = charset_normalizer.detect(rawdata.read(10000))
print(result)
```

```python
# read in the file with the encoding detected by charset_normalizer
ks2016_data = pd.read_csv(filename, encoding='Windows-1252', low_memory=False)
ks2016_data.head()
```
