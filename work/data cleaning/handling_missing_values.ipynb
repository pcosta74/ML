```python
import pandas as pd
import numpy as np

# set seed for reproducibility
np.random.seed(0) 

# Load data
nfl_data = pd.read_csv("../datasets/nflplaybyplay2009to2016/NFL Play by Play 2009-2017 (v4).csv")
nfl_data.head(10)
```

```python
### Approach #1: drop missing values
nfl_data.dropna()
```

```python
# get the number of missing data points per column
missing_values_count = nfl_data.isnull().sum()
missing_values_count[:10]
```

```python
# how many total missing values do we have?
total_cells = np.prod(nfl_data.shape)
total_missing = missing_values_count.sum()

# percent of data that is missing
percent_missing = (total_missing/total_cells) * 100
print(f"Missing data percentage: {percent_missing:.3f}")
```

```python
# remove all columns with at least one missing value
columns_with_na_dropped = nfl_data.dropna(axis=1)
columns_with_na_dropped.head()
```

```python
# just how much data did we lose?
print("Columns in original dataset: %d" % nfl_data.shape[1])
print("Columns with na's dropped: %d" % columns_with_na_dropped.shape[1])
```
