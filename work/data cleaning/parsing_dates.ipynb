```python
import datetime
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

%matplotlib inline
np.random.seed(0)
```

```python
landslides_data = pd.read_csv("../datasets/landslide-events/catalog.csv")
landslides_data.head()
```

```python
# print the first few rows of the date column
print(landslides_data['date'].head())
```

```python
# check the data type of our date column
landslides_data['date'].dtype
```

```python
# create a new column, date_parsed, with the parsed dates
landslides_data['date_parsed'] = pd.to_datetime(landslides_data['date'], format="%m/%d/%y")
# print the first few rows
landslides_data['date_parsed'].head()
```

```python
# get the day of the month from the date_parsed column
day_of_month_landslides = landslides_data['date_parsed'].dt.day
day_of_month_landslides.head()
```

```python
# remove na's
day_of_month_landslides = day_of_month_landslides.dropna()

# plot the day of the month
sns.displot(day_of_month_landslides, kde=False, bins=31)
plt.show()
```
